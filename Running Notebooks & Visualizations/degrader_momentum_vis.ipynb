```python
import numpy as np
import matplotlib.pyplot as plt
```

This plots how different degraders affect the momentum loss.

```python
mu_data = np.loadtxt('degrader_momentum_results_mu+.csv', delimiter=',')
e_data = np.loadtxt('degrader_momentum_results_e+.csv',delimiter=",")
```

```python
#convert the resultant momenta into percent changes.
new_col = (1 - mu_data[:, 3] / mu_data[:, 2]).reshape(-1, 1)
mu_result = np.hstack((mu_data, new_col))

new_col = (1 - e_data[:, 3] / e_data[:, 2]).reshape(-1, 1)
e_result = np.hstack((e_data,new_col))
```

```python
cats = mu_result[:, 0].astype(int)
x = mu_result[:, 1]
y = mu_result[:, 4]

cat_dict = {
    0:"Be",
    1:"Pb",
    2:"Al",
    3:"Au"
}

colors = ['red', 'green', 'blue', 'purple']

for c in np.unique(cats):
    mask = cats == c
    plt.scatter(x[mask], y[mask], color=colors[c], label=f'{cat_dict[c]}')

plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.show()
```
