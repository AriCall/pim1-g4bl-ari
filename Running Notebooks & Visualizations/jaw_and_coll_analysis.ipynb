```python
def ifp_step_panel(muon_dfs,pos_dfs):
    fig, ax = plt.subplots(1,5,figsize=(18,5))
    ax[0].scatter(muon_dfs['NTuple/Z10620;1']['x'],muon_dfs['NTuple/Z10620;1']['Pz'],label='Muons',color='blue',alpha=0.5)
    ax[0].scatter(pos_dfs['NTuple/Z10620;1']['x'],pos_dfs['NTuple/Z10620;1']['Pz'],label='Positrons',color='orange',alpha=0.5)
    ax[0].legend()
    ax[0].set_title('Before IFP')
    ax[0].set_ylabel('Pz')
    ax[0].set_xlabel('x position')
    ax[0].set_ylim(90,105)
    ax[0].set_xlim(-150,150)

    ax[1].scatter(muon_dfs['NTuple/Z11475;1']['x'],muon_dfs['NTuple/Z11475;1']['Pz'],label='Muons',color='blue',alpha=0.5)
    ax[1].scatter(pos_dfs['NTuple/Z11475;1']['x'],pos_dfs['NTuple/Z11475;1']['Pz'],label='Positrons',color='orange',alpha=0.5)
    ax[1].legend()
    ax[1].set_title('After Collimator')
    ax[1].set_ylabel('Pz')
    ax[1].set_xlabel('x position')  
    ax[1].set_ylim(90,105)
    ax[1].set_xlim(-150,150)

 

    ax[2].scatter(muon_dfs['NTuple/Z12140;1']['x'],muon_dfs['NTuple/Z12140;1']['Pz'],label='Muons',color='blue',alpha=0.5)
    ax[2].scatter(pos_dfs['NTuple/Z12140;1']['x'],pos_dfs['NTuple/Z12140;1']['Pz'],label='Positrons',color='orange',alpha=0.5)
    ax[2].legend()
    ax[2].set_title('After Degrader')
    ax[2].set_ylabel('Pz')
    ax[2].set_xlabel('x position')  
    ax[2].set_ylim(85,105)
    ax[2].set_xlim(-150,150)

 
    ax[3].scatter(muon_dfs['NTuple/Z16963;1']['x'],muon_dfs['NTuple/Z16963;1']['Pz'],label='Muons',color='blue',alpha=0.5)
    ax[3].scatter(pos_dfs['NTuple/Z16963;1']['x'],pos_dfs['NTuple/Z16963;1']['Pz'],label='Positrons',color='orange',alpha=0.5)
    ax[3].legend()
    ax[3].set_title('After Turn')
    ax[3].set_ylabel('Pz')
    ax[3].set_xlabel('x position')  
    ax[3].set_ylim(85,105)
    ax[3].set_xlim(-150,150)

    ax[4].legend()
    ax[4].set_title('At Target')
    ax[4].set_ylabel('Pz')
    ax[4].set_xlabel('x position')  
    ax[4].set_ylim(85,105)
    ax[4].set_xlim(-150,150)
    ax[4].scatter(muon_dfs['NTuple/Z22159;1']['x'],muon_dfs['NTuple/Z22159;1']['Pz'],label='Muons',color='blue',alpha=0.5)
    ax[4].scatter(pos_dfs['NTuple/Z22159;1']['x'],pos_dfs['NTuple/Z22159;1']['Pz'],label='Positrons',color='orange',alpha=0.5)
 


ifp_step_panel(muon_dfs,pos_dfs)
```
