```python
#Find the degraders that have 1% momentum loss across them

data = np.loadtxt('au_ag_cu_momentum_loss_new.csv',delimiter=',')
au = data[0:50]
ag = data[51:100]
cu = data[101:150]
w = data[151:200]

plt.plot(au[:,1], au[:,-2]*100, label='Au', color='blue', linestyle='solid')
plt.plot(ag[:,1], ag[:,-2]*100, label='Ag', color='red', linestyle='solid')
plt.plot(cu[:,1], cu[:,-2]*100, label='Cu', color='green', linestyle='solid')
plt.plot(w[:,1], w[:,-2]*100, label='W', color='purple', linestyle='solid')

plt.title('Momentum Loss through Material - Mean')
plt.xlabel('Material Length (mm)')
plt.ylabel('% Momentum After Degrader')
plt.axhline(-1,0,1, color="purple")

plt.legend() 
plt.grid(True)
plt.show()
print(w[:,1], w[:,-2]*100)
```

```python
data = np.loadtxt('degrader_plastic_100.csv',delimiter=',')
length = data[:,0]
mu_pz_start=data[:,1]
mu_pz_end=data[:,2]
pos_pz_start=data[:,3]
pos_pz_end=data[:,4]
mucount = data[:,5]
poscount = data[:,6]
```

```python
mu_energy_loss = mu_pz_end/mu_pz_start
pos_energy_loss = pos_pz_end/pos_pz_start
```

```python
#Copper
plt.plot(length, mu_energy_loss, label='Muon', color='blue', linestyle='solid')
plt.plot(length, pos_energy_loss, label='Positron', color='red', linestyle='solid')

plt.title('energy loss for plastic')
plt.xlabel('mm, plastic')
plt.ylabel('energy remaining')
#plt.axhline(99,0,1, color="purple")

plt.legend() 
plt.grid(True)
plt.show()
print(length,mu_energy_loss)
#there is 1% separation at 2.88 mm
```

```python
plt.plot(length, mucount, label='Muon', color='blue', linestyle='solid')
plt.plot(length, poscount, label='Positron', color='red', linestyle='solid')

plt.title('number of particles at end')
plt.xlabel('mm, plastic')
plt.ylabel('count remaining')
#plt.axhline(99,0,1, color="purple")

plt.legend() 
plt.grid(True)
plt.show()
```

```python
plt.plot(length, mucount/(mucount+poscount)*100, label='%muon', color='blue', linestyle='solid')

plt.title('percent muon at end')
plt.xlabel('mm, plastic')
plt.ylabel('percent')
#plt.axhline(99,0,1, color="purple")

plt.legend() 
plt.grid(True)
plt.show()
print(length)
print(mucount/(mucount+poscount)*100)
```
